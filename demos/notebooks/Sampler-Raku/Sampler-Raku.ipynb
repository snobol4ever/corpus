# Sampler-Raku

Four small Raku programs from Rosetta Code, one section each: a lazy sequence, a string rewriter, a drawing and a recursive encoder. Raku has its own kernel entry, so a cell needs no tag. Run the cells in order with Run All, then change a number and run a section again. Hover a name for what it is, press Tab for completions, and set a breakpoint in the gutter to step through a routine with the debugger.

## Pascal's triangle

A lazy, infinite sequence in one line. Each row is the previous row with a zero on one end added to the same row with a zero on the other end: `Z+` zips two lists and adds them pairwise. The `...` operator keeps generating rows, and `^10` takes the first ten.

```raku
constant @pascal = [1], -> $prev { [0, |$prev Z+ |$prev, 0] } ... *;

.say for @pascal[^10];
```

## Look and say

Each term describes the one before it: `1` is one 1, so the next term is `11`; that is two 1s, so `21`; one 2 and one 1, so `1211`. A regex matches a character with its repeats (`(.)$0*`), and `subst` with `:g` replaces every run by its length and character.

```raku
.say for ('1', *.subst(/(.)$0*/, { .chars ~ .[0] }, :g) ... *)[^15];
```

## Sierpinski triangle

Start with one star. Each round puts a copy of the picture above, centred, and two copies side by side below it. Four rounds draw a triangle of holes.

```raku
sub sierpinski ($n) {
    my @down  = '*';
    my $space = ' ';
    for ^$n {
        @down = |("$space$_$space" for @down), |("$_ $_" for @down);
        $space x= 2;
    }
    return @down;
}

.say for sierpinski 4;
```

## Roman numerals

A `multi` sub has two bodies: `roman(0)` is the empty string, and `roman(Int $n)` peels off the biggest symbol it can and calls itself on the rest, so the numeral for 1994 is built one symbol at a time.

```raku
my %symbols =
    1 => "I", 5 => "V", 10 => "X", 50 => "L", 100 => "C",
    500 => "D", 1_000 => "M";

my @subtractors =
    1_000, 100,  500, 100,  100, 10,  50, 10,  10, 1,  5, 1,  1, 0;

multi sub roman (0) { '' }
multi sub roman (Int $n) {
    for @subtractors -> $cut, $minus {
        $n >= $cut
            and return %symbols{$cut} ~ roman($n - $cut);
        $n >= $cut - $minus
            and return %symbols{$minus} ~ roman($n + $minus);
     }
}

# Sample usage: the full program counts 1 to 2010; here are a few

for 1, 4, 9, 14, 40, 90, 400, 1994, 2010 -> $x {
    say roman($x);
}
```
