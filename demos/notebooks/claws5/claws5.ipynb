# Word Tagger
In a tagged corpus every word carries its part of speech: `power_NN1` is a singular noun, `taxing_VVG` a verb ending in -ing. This notebook reads a small piece of a real tagged corpus, `claws5.in` (1077 bytes), with **one Snocone pattern**. An action inside the pattern, `token()`, files each word under its sentence number and its tag at the moment the pattern matches it. Run the cells in order.

## 1. The pattern, and the run
The pattern reads a number or a word with its tag, over and over, to the end of the file. The last lines open `claws5.in`, beside this notebook, and match the whole of it. The whole corpus, `CLAWS5inTASA.dat` (66757 bytes), stands beside it too: to read it instead, edit the one `INPUT(` line of the first code cell below and change `'claws5.in [-r200000]'` to `'CLAWS5inTASA.dat [-r200000]'`, which the comment on that line spells out, then run the cells again.

```snocone
// claws5.sc -- CLAWS5 POS-tagged corpus tokenizer (Snocone).
// ENG 685, Lon Cherryholmes Sr.
// Equivalent of claws5.sno; same output, same oracle ref (claws5.ref).
// Run: scrip claws5.sc < claws5.input
//------------------------------------------------------------------------------
function token() {
    num                 = +num;
    if (IDENT(mem))                  mem                 = TABLE();
    if (IDENT(mem[num]))             mem[num]            = TABLE();
    if (IDENT(mem[num][wrd]))        mem[num][wrd]       = TABLE();
    if (IDENT(mem[num][wrd][tag]))   mem[num][wrd][tag]  = 0;
    mem[num][wrd][tag]  = mem[num][wrd][tag] + 1;
    token               = .dummy;
    nreturn;
}
//------------------------------------------------------------------------------
&TRIM   =   0;
claws   =   POS(0)
                ARBNO(
                  FENCE(
                    (SPAN('0123456789') . num) '_CRD :_PUN'
                  | (NOTANY('_') BREAK('_')) . wrd
                    '_'
                    (ANY(&UCASE) SPAN('0123456789' &UCASE)) . tag . *token()
                  )
                  SPAN(' ' CHAR(10))
                )
                RPOS(0);
INPUT(.INPUT, 9, 'claws5.in [-r200000]');  // or 'CLAWS5inTASA.dat [-r200000]', the full file (66757 B)
if (src = INPUT) {
    if (src ? claws) OUTPUT = 'matched bytes=' SIZE(src);
    else             OUTPUT = 'Pattern match failed';
} else               OUTPUT = 'Pattern match failed';
```

## 2. What it found
The words are now in a table of tables, `mem`, by sentence, word and tag. Here is the first sentence, its words in alphabetical order.

```snocone
w = SORT(mem[1]);
PROTOTYPE(w) ? BREAK(',') . n;
OUTPUT = n ' different words in the first sentence; the first few, with their tags:';
for (i = 1; LE(i, 6); i = i + 1) {
    t = SORT(w[i,2]);
    OUTPUT = w[i,1] ' is tagged ' t[1,1];
}
OUTPUT = num ' sentences in all';
```
