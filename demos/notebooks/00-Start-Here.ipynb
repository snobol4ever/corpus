# Start here: SCRIP in Jupyter

SCRIP runs SNOBOL4, Snocone, Icon, Prolog, Pascal and Raku from one kernel, and in the mixed notebooks those languages call each other in the same document.
Click a name to open it, then use **Run All** (or Shift+Enter one cell at a time). Every notebook opens with its outputs empty, so what you see is SCRIP running in front of you.


## The wow: languages working together

* [Family-Tree-Ask-and-Draw](../scriptix/mixes/notebooks/Family-Tree-Ask-and-Draw.ipynb): ask a family tree in plain English and see the answer drawn as a picture. SNOBOL4 reads the question, Prolog finds the answer, and the picture is drawn in the notebook.
* [Family-Tree-Ask-in-English](../scriptix/mixes/notebooks/Family-Tree-Ask-in-English.ipynb): the same family tree, answered in words.
* [Calculator-Three-Languages](../scriptix/mixes/notebooks/Calculator-Three-Languages.ipynb): one calculator built from three languages that call each other.
* [JSON-Round-Trip](../scriptix/mixes/notebooks/JSON-Round-Trip.ipynb): SNOBOL4 reads JSON and Icon writes it back.
* [Library-Desk](../scriptix/mixes/notebooks/Library-Desk.ipynb): a library desk built in four languages.
* [Spreadsheet](../scriptix/mixes/notebooks/Spreadsheet.ipynb): a small spreadsheet built in three languages.
* [Trip-Planner](../scriptix/mixes/notebooks/Trip-Planner.ipynb): plan a trip with a search in one language and a report in another.
* [Log-Detective](../scriptix/mixes/notebooks/Log-Detective.ipynb): read a server log and find what went wrong.
* [Calculus-Homework](../scriptix/mixes/notebooks/Calculus-Homework.ipynb): differentiate and simplify formulas.
* [Sentence-Maker-and-Reader](../scriptix/mixes/notebooks/Sentence-Maker-and-Reader.ipynb): Icon writes sentences and SNOBOL4 reads them back.
