# Start here: SCRIP in Jupyter

SCRIP runs Snocone, Icon, Prolog, Pascal and Raku from one kernel, and in the mixed notebooks those languages call each other in the same document.
Click a name to open it, then use **Run All** (or Shift+Enter one cell at a time). Every notebook opens with its outputs empty, so what you see is SCRIP running in front of you.


## The wow: languages working together

* [Family-Tree-Ask-and-Draw](Family-Tree-Ask-and-Draw/Family-Tree-Ask-and-Draw.ipynb): ask a family tree in plain English and see the answer drawn as a picture. Snocone reads the question, Prolog finds the answer, and the picture is drawn in the notebook.
* [Family-Tree-Ask-in-English](Family-Tree-Ask-in-English/Family-Tree-Ask-in-English.ipynb): the same family tree, answered in words.
* [Calculator-Three-Languages](Calculator-Three-Languages/Calculator-Three-Languages.ipynb): one calculator built from three languages that call each other.
* [JSON-Round-Trip](JSON-Round-Trip/JSON-Round-Trip.ipynb): Snocone reads JSON and Icon writes it back.
* [Library-Desk](Library-Desk/Library-Desk.ipynb): a library desk built in four languages.
* [Spreadsheet](Spreadsheet/Spreadsheet.ipynb): a small spreadsheet built in three languages.
* [Trip-Planner](Trip-Planner/Trip-Planner.ipynb): plan a trip with a search in one language and a report in another.
* [Log-Detective](Log-Detective/Log-Detective.ipynb): read a server log and find what went wrong.
* [Calculus-Homework](Calculus-Homework/Calculus-Homework.ipynb): differentiate and simplify formulas.
* [Sentence-Maker-and-Reader](Sentence-Maker-and-Reader/Sentence-Maker-and-Reader.ipynb): Icon writes sentences and Snocone reads them back.


## Snocone: patterns that parse

Snocone is SCRIP's language for pattern matching: grammars written as patterns and matched against real text. It is the default language of the mixed notebooks.

* [calculator](calculator/calculator.ipynb): a calculator whose whole grammar is one pattern.
* [json](json/json.ipynb): a JSON parser reading a real document.
* [treebank](treebank/treebank.ipynb): reads a treebank of parsed sentences and counts the noun and verb phrases.
* [claws5](claws5/claws5.ipynb): reads a corpus tagged with parts of speech.
