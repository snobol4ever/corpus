**6.4** A pattern is built from simpler patterns: this one matches `MATCH` with an optional suffix `S` or `ES`.

```scrip
        ROOT = 'MATCH'
        OUTPUT = test('MATCHES', POS(0) ROOT ('S' | 'ES' | '') RPOS(0))
```

A pattern can also test a counter: `FOX` matches only when `N` is 1, `WOLF` only when `N` is 2.

```scrip
        PAT = *EQ(N,1) 'FOX' | *EQ(N,2) 'WOLF'
        N = 0
n12     N = LT(N,2) N + 1                                       :F(e12)
        OUTPUT = N ' FOX ' test('FOX', PAT)
        OUTPUT = N ' WOLF ' test('WOLF', PAT)                   :(n12)
e12
```

Alternatives nest: three words that share a beginning and an end.

```scrip
        COMP = POS(0) 'COMP' ('AT' | 'RE' ('HEN' | 'S') 'S') 'IBLE' RPOS(0)
        OUTPUT = test('COMPATIBLE', COMP)
        OUTPUT = test('COMPREHENSIBLE', COMP)
        OUTPUT = test('COMPRESSIBLE', COMP)
```
