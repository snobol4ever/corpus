# Snocone in SCRIP: the core rungs

### fence_replace_1

```scrip
%%snocone
/* TWIN IS AN EQUIVALENCE, NOT A TRANSLITERATION: SNOBOL4 has no if-block, so the twin expresses the same branch with a goto label; the guarded assignment and the printed result are identical */
s = 'xxabc';
r = 'no';
if (s ? FENCE 'abc') { r = 'yes'; }
OUTPUT = r;
```

### ladder_prog__pat_fence

```scrip
%%snocone
if ('1AB+' ? ANY('AB') FENCE '+') { OUTPUT = 'wrong'; } else { OUTPUT = 'fence-blocks-ok'; }
if ('A+B'  ? ANY('AB') FENCE '+') { OUTPUT = 'fence-pass-ok'; }
```

### arb_capture_1

```scrip
%%snocone
s = 'abc';
s ? ARB . t 'c';
OUTPUT = t;
```

### arbno_capture_1

```scrip
%%snocone
s = 'ababX';
s ? ARBNO('ab') . t 'X';
OUTPUT = t;
```

### bal_capture_1

```scrip
%%snocone
s = '(a+b)';
s ? BAL . t;
OUTPUT = t;
```

### ladder_prog__pat_immed

```scrip
%%snocone
s = 'abcdef';
s ? LEN(2) $ first;
OUTPUT = first;
```

### notany_capture_1

```scrip
%%snocone
s = 'xyz';
s ? NOTANY('y') . t;
OUTPUT = t;
```

### rtab_capture_1

```scrip
%%snocone
s = 'hello';
s ? RTAB(2) . t;
OUTPUT = t;
```

### tab_capture_1

```scrip
%%snocone
s = 'hello';
s ? TAB(3) . t;
OUTPUT = t;
```

### abort_alt_1

```scrip
%%snocone
s = 'abc';
OUTPUT = 'before';
s ? ('x' | ABORT) 'c';
OUTPUT = 'after';
```

### ladder_prog__datatype

```scrip
%%snocone
OUTPUT = DATATYPE('hello');
OUTPUT = DATATYPE(42);
OUTPUT = DATATYPE(3.14);
OUTPUT = CONVERT('42', 'INTEGER') + 1;
```
