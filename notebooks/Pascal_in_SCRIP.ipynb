### ladder__rung04_for_exit_value_is_the_final_value

### program_function_procedure_5

```scrip
%%pascal
program goto3;
var acc: integer;

procedure cnt(k: integer);
label 1;
var c: integer;
begin
  c := 0;
1:
  c := c + 1;
  if c < k then goto 1;
  acc := acc + c
end;

function dbl(x: integer): integer;
label 1;
var s: integer;
begin
  s := 0;
1:
  s := s + x;
  x := x - 1;
  if x > 0 then goto 1;
  dbl := s
end;

begin
  acc := 0;
  cnt(4);
  cnt(7);
  writeln(acc);
  writeln(dbl(5))
end.
```

### ladder__rung16_real_const_fold_bound

```scrip
%%pascal
program constfoldbound;
{$mode delphi}
const
  M = 100000;
  C = trunc(M / (ln(M) - 1.08)) + 100;
  R = round(2.5);
  S = trunc(sqrt(16.0));
  E = trunc(exp(2));
var
  a : array[0..C] of integer;
  b : array[1..R] of char;
  d : array[S..S + E] of byte;
begin
  a[C] := 7;
  writeln(C, ' ', high(a), ' ', a[high(a)]);
  writeln(R, ' ', low(b), ' ', high(b));
  writeln(S, ' ', low(d), ' ', high(d), ' ', length(d));
end.
```

### program_array_while_replace_1

```scrip
%%pascal
(*$c+,d+,l-,t-*)
(*
  Sieve of Eratosthenes. Displays primes less than 'n'.
  
  By Christophe Staiesse.
*)
program eratosthenessieve;
const
        n = 100;
var
        sv: array[2..n] of boolean;
        i,j: integer;
begin
        for i := 2 to n do sv[i] := true;
        i := 2;
        while sqr(i) <= n do begin
                if sv[i] then begin
                        j := sqr(i);
                        while j <= n do begin
                                sv[j] := false;
                                j := j+i; 
                        end;
                end;
                i := i +1;
        end;
        for i := 2 to n do
                if sv[i] then writeln(i);
end.
```
