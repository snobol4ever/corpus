# Unicon in SCRIP

Unicon adds classes to Icon: an account keeps a log of its deposits and refusals. Run All, hover a name, press Tab for a method, Alt-click a call to jump to its definition, or put a breakpoint in the gutter and step into a method.

```unicon
class Account (balance, log)
  method deposit (amt)
    balance +:= amt;
    put (log, "deposit " || amt);
    return balance;
  end
  method withdraw (amt)
    if balance >= amt then { balance -:= amt; return balance; };
    put (log, "refused " || amt);
    fail;
  end
  initially (b)
    balance := b;
    log := [];
end
```

```unicon
procedure main ()
   local a;
   a := Account (100);
   a.deposit (25);
   a.withdraw (500);
   write (a.withdraw (50));
   every write (!a.log);
end
```
