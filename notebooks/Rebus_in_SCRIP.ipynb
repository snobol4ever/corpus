# Rebus in SCRIP: the core rungs

### alt_replace_1

```scrip
%%rebus
function main()
if "cat" ? ("cat" | "category") then OUTPUT := "matched" else OUTPUT := "no match"
if "category" ? ("cat" | "category") then OUTPUT := "matched" else OUTPUT := "no match"
end
```

### alt_replace_2

```scrip
%%rebus
function main()
if "banana" ? ("apple" | "banana" | "cherry") then OUTPUT := "matched" else OUTPUT := "no match"
if "kiwi" ? ("apple" | "banana" | "cherry") then OUTPUT := "matched" else OUTPUT := "no match"
end
```

### len_capture_2

```scrip
%%rebus
function main()
"hello world" ? (LEN(5) . w)
OUTPUT := w
end
```

### break_rem_capture_1

```scrip
%%rebus
function main()
"key=value" ? (BREAK("=") . k & "=" & REM . v)
OUTPUT := k
OUTPUT := v
end
```

### break_rem_capture_2

```scrip
%%rebus
function main()
s := "key=value"
s ? (BREAK("=") . k & "=" & REM . v) <- (k & ":" & v)
OUTPUT := s
end
```

### arb_break_rem_1

```scrip
%%rebus
function main()
"foo(bar)baz" ? (BREAK("(") . pre & "(" & ARB . inner & ")" & REM . post)
OUTPUT := pre
OUTPUT := inner
OUTPUT := post
end
```
