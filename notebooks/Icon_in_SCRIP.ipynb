# Icon in SCRIP

Icon runs here as a kernel of its own: procedures, recursion and lists, one cell a piece. This is the Towers of Hanoi: three needles, N disks, move the pile from A to C by way of B. Run the cells in order with Run All. Hover a name for what it is, press Tab for the procedures you have defined, Alt-click a call to jump to its procedure, and set a breakpoint in the gutter to step through the recursion with the debugger.

```icon
global moves, log
procedure setup()
   moves := 0;
   log := [];
end
```

```icon
procedure move(n, src, dst)
   moves +:= 1;
   put(log, src || ">" || dst);
   write("Move disk ", n, " from ", src, " to ", dst);
   return;
end
procedure hanoi(n, src, dst, via)
   if n = 0 then return;
   hanoi(n - 1, src, via, dst);
   move(n, src, dst);
   hanoi(n - 1, via, dst, src);
end
```

```icon
procedure main()
   local k;
   setup();
   hanoi(3, "A", "C", "B");
   write("moves: ", moves);
   every k := 1 to *log do writes(log[k], if k < *log then "," else "\n");
end
```
