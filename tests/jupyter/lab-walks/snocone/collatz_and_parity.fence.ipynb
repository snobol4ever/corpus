# Collatz steps and mutual recursion in Snocone

A whole program in three cells: a loop that follows the Collatz rule, two functions that call each other to decide parity, and a run that uses both.

```snocone
function Collatz(n, c) {
    c = 0;
    while (NE(n, 1)) {
        if (EQ(REMDR(n, 2), 0)) { n = n / 2; } else { n = 3 * n + 1; }
        c = c + 1;
    }
    Collatz = c; return;
}
function IsEven(n) {
    if (EQ(n, 0)) return;
    if (IsOdd(n - 1)) return;
    freturn;
}
function IsOdd(n) {
    if (EQ(n, 0)) freturn;
    if (IsEven(n - 1)) return;
    freturn;
}
```

```snocone
i = 1;
while (LE(i, 10)) {
    OUTPUT = i ' takes ' Collatz(i) ' steps';
    i = i + 1;
}
```

```snocone
%%snocone
best = 0;
at = 0;
i = 1;
while (LT(i, 30)) {
    c = Collatz(i);
    if (GT(c, best)) { best = c; at = i; }
    i = i + 1;
}
OUTPUT = 'longest below 30: ' at ' with ' best ' steps';
if (IsEven(6)) { OUTPUT = '6 is even'; }
if (IsOdd(7)) { OUTPUT = '7 is odd'; }
```
