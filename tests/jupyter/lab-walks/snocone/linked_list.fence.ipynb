# A linked list in Snocone

A whole program in three cells: a struct for a list cell, functions that build, measure, print and reverse a list, and a run that uses them.

```snocone
struct node { val, link }
function Push(l, v) { Push = node(v, l); return; }
function Length(l, c) {
    c = 0;
    while (DIFFER(l)) { c = c + 1; l = link(l); }
    Length = c; return;
}
function Sum(l, s) {
    s = 0;
    while (DIFFER(l)) { s = s + val(l); l = link(l); }
    Sum = s; return;
}
function Show(l, s) {
    s = '';
    while (DIFFER(l)) { s = s val(l) ' '; l = link(l); }
    Show = s; return;
}
function Reverse(l, r, nx) {
    while (DIFFER(l)) { nx = link(l); link(l) = r; r = l; l = nx; }
    Reverse = r; return;
}
```
