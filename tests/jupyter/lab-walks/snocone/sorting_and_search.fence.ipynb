# Sorting and binary search in Snocone

A whole program in three cells: an array filled by a small generator, an insertion sort and a binary search written as functions that call each other and walk the array.

```snocone
function Fill(a, n, i, x) {
    x = 5;
    i = 1;
    while (LE(i, n)) { x = REMDR(x * 7 + 3, 31); a[i] = x; i = i + 1; }
}
function Show(a, n, i, s) {
    s = a[1];
    i = 2;
    while (LE(i, n)) { s = s ' ' a[i]; i = i + 1; }
    Show = s; return;
}
function Sort(a, n, i, j, t) {
    i = 2;
    while (LE(i, n)) {
        t = a[i];
        j = i - 1;
        while (GT(j, 0)) {
            if (LE(a[j], t)) { break; }
            a[j + 1] = a[j];
            j = j - 1;
        }
        a[j + 1] = t;
        i = i + 1;
    }
}
function Find(a, n, x, lo, hi, mid) {
    lo = 1; hi = n;
    while (LE(lo, hi)) {
        mid = (lo + hi) / 2;
        if (EQ(a[mid], x)) { Find = mid; return; }
        if (LT(a[mid], x)) { lo = mid + 1; } else { hi = mid - 1; }
    }
    freturn;
}
```

```snocone
n = 8;
a = ARRAY(n);
Fill(a, n);
OUTPUT = 'before: ' Show(a, n);
Sort(a, n);
OUTPUT = 'after:  ' Show(a, n);
```

```snocone
%%snocone
x = a[3];
OUTPUT = 'find ' x ' at ' Find(a, n, x);
if (Find(a, n, 99)) { OUTPUT = 'find 99 found'; } else { OUTPUT = 'find 99 not found'; }
```
