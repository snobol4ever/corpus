# Counting words with patterns in Snocone

A whole program in three cells: functions that take a sentence apart with patterns, a table of word counts, and a report of the words that repeat.

```snocone
function NextWord(w) {
    text ? POS(0) SPAN(' ') = '';
    if (IDENT(text)) freturn;
    if (text ? POS(0) BREAK(' ') . w = '') { NextWord = w; return; }
    NextWord = text;
    text = '';
    return;
}
function Count(words, w, n) {
    while (w = NextWord()) {
        n = words[w];
        if (IDENT(n)) { n = 0; }
        words[w] = n + 1;
    }
}
```

```snocone
text = 'the cat and the dog and the bird sat';
words = TABLE();
Count(words);
```

```snocone
%%snocone
a = SORT(words);
i = 1;
while (a[i, 1]) {
    if (GT(a[i, 2], 1)) { OUTPUT = a[i, 1] ' ' a[i, 2]; }
    i = i + 1;
}
```
