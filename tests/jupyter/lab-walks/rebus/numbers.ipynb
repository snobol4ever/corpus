```rebus
function fact(n)
  if n <= 1 then return 1
  return n * fact(n - 1)
end

function fib(n)
  if n < 2 then return n
  return fib(n - 1) + fib(n - 2)
end
```

```rebus
function isprime(n)
  local d
  if n < 2 then fail
  d := 2
  while d * d <= n do {
    if n % d = 0 then fail
    d := d + 1
  }
  return n
end

function gcd(a, b)
  if b = 0 then return a
  return gcd(b, a % b)
end
```
