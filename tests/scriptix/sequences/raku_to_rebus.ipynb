```scrip
%%rebus
function main()
  OUTPUT := "size " || PROTOTYPE(SEQ)
  OUTPUT := "second " || SEQ[2]
  BACK := ARRAY(3)
  BACK[1] := 40
  BACK[2] := 50
  BACK[3] := 60
end
```

```scrip
%%raku
our $BACK;
say "back size {$BACK.elems}";
say "back second {$BACK[1]}";
```
