```scrip
%%rebus
record tree(t, v, n, c)
function main()
  OUTPUT := "root " || t(ROOT) || " " || n(ROOT)
  K := c(ROOT)
  OUTPUT := "leaves " || v(K[1]) || " " || v(K[2])
  J := ARRAY(2)
  J[1] := tree("num", "3", 0)
  J[2] := tree("num", "4", 0)
  BACK := tree("times", "*", 2, J)
end
```
